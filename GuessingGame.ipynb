```python
import random
import gradio as gr

def generate_secret(start, end):
  return random.randint(int(start), int(end))
def start_game(start, end, attempts):
  secret = generate_secret(start, end)
  attempts_left = int(attempts)
  message = f"Game started! Guess a number between {int(start)} and {int(end)}."
  history = ""
  return secret, attempts_left, message, history
def evaluate_guess(guess, secret):
  if guess < secret:
      return "Too low"
  elif guess > secret:
      return "Too high"
  else:
      return "Correct"
def update_history(history, guess):
  if history == "":
      return str(int(guess))
  return history + ", " + str(int(guess))
def process_guess(guess, secret, attempts_left, history):
  guess = int(guess)
  secret = int(secret)
  attempts_left = int(attempts_left)

  result = evaluate_guess(guess, secret)
  history = update_history(history, guess)

  if result == "Correct":
      message = f"Correct! You guessed the secret number {secret}."
      return secret, attempts_left, message, history

  attempts_left -= 1

  if attempts_left == 0:
      message = f"Game over! You ran out of attempts. The secret number was {secret}."
  else:
      message = f"{result}! Attempts left: {attempts_left}"

  return secret, attempts_left, message, history

# ----- GUI (DO NOT MODIFY) -----
with gr.Blocks() as demo:
  gr.Markdown("# Number Guessing Game")

  secret_state = gr.State(0)
  attempts_state = gr.State(0)
  history_state = gr.State("")

  with gr.Row():
    start_input = gr.Number(label="Range Start", value=1, precision=0)
    end_input = gr.Number(label="Range End", value=20, precision=0)
    attempts_input = gr.Number(label="Attempts", value=5, precision=0)

  start_button = gr.Button("Start Game")

  status_box = gr.Textbox(label="Game Status")
  history_box = gr.Textbox(label="Guess History")

  with gr.Row():
    guess_input = gr.Number(label="Your Guess", precision=0)
    guess_button = gr.Button("Submit Guess")

  start_button.click(
    fn=start_game,
    inputs=[start_input, end_input, attempts_input],
    outputs=[secret_state, attempts_state, status_box, history_state]
  )
  guess_button.click(
    fn=process_guess,
    inputs=[guess_input, secret_state, attempts_state, history_state],
    outputs=[secret_state, attempts_state, status_box, history_box]
  )
demo.launch()
```
