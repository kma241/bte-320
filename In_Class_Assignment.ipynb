```python
import math

def calculate_eoq(demand, order_cost, holding_cost):
    if demand <= 0 or order_cost <= 0 or holding_cost <= 0:
        raise ValueError("All inputs must be positive numbers.")
    return math.sqrt((2 * demand * order_cost) / holding_cost)

def eoq_main():
    try:
        demand = float(input("Enter annual demand (units/year): "))
        order_cost = float(input("Enter ordering cost ($/order): "))
        holding_cost = float(input("Enter annual holding cost per unit ($/year/unit): "))

        eoq = calculate_eoq(demand, order_cost, holding_cost)
        print(f"Economic Order Quantity (EOQ): {eoq:.2f} units")

    except ValueError as e:
        print(f"Error: {e}")

eoq_main()
```

```python
def calcFee(hours):
  base_fee = 5.0
  hourly_rate = 2.50
  min_fee = 10.0
  max_fee = 20.0
  fee = base_fee + (hourly_rate * hours)
  if fee < min_fee:
    fee = min_fee
  elif fee > max_fee:
    fee = max_fee
  return fee

def main():
  try:
    hours_input = int(input("Enter number of parking hours:"))
    if hours_input < 0:
      print("Parking hours cannot be negative. Please enter a positive number.")
      return
    parking_fee = calcFee(hours_input)
    print(f"Parking fee for {hours_input} hours: ${parking_fee:.2f}")
  except ValueError:
    print("Invalid input. Please enter a valid number of hours.")

main()
```

```python
def roi(cost, revenue):
    assert isinstance(cost, (int, float))
    assert isinstance(revenue, (int, float))
    assert cost != 0

    return (revenue - cost) / cost


def loan_payment(principal, annual_rate, months):
    assert isinstance(principal, (int, float)), "principal must be numeric"
    assert isinstance(annual_rate, (int, float)), "annual_rate must be numeric"
    assert isinstance(months, (int, float)), "months must be numeric"

    if principal <= 0:
        raise ValueError("principal must be greater than 0")
    if months <= 0:
        raise ValueError("months must be greater than 0")
    if annual_rate < 0:
        raise ValueError("annual_rate cannot be negative")

    r = annual_rate / 12

    if r != 0:
        pmt = r * principal / (1 - (1 + r) ** (-months))
    else:
        pmt = principal / months

    return pmt


try:
    cost = float(input("Enter cost: "))
    revenue = float(input("Enter revenue: "))
    print("ROI:", roi(cost, revenue))

    principal = float(input("Enter principal: "))
    annual_rate = float(input("Enter annual rate (decimal): "))
    months = float(input("Enter number of months: "))
    print("Monthly payment:", loan_payment(principal, annual_rate, months))

except ValueError as ve:
    print("ValueError:", ve)
except AssertionError as ae:
    print("AssertionError:", ae)
except Exception as e:
    print("Unexpected error:", e)
```

```python
def getValues(n):
    if not isinstance(n, int) or n <= 0:
        raise ValueError("n must be a positive integer.")

    values = []
    for i in range(n):
        value = float(input(f"Enter value {i+1}: "))
        values.append(value)

    return values

def calcAverage(alist):
    if not isinstance(alist, list):
        raise TypeError("Input must be a list.")

    if len(alist) <= 3:
        raise ValueError("List must contain more than 3 values to exclude the first three.")

    trimmed_list = alist[3:]
    return sum(trimmed_list) / len(trimmed_list)

def main():
    try:
        n = int(input("Enter number of values: "))
        values = getValues(n)
        avg = calcAverage(values)
        print(f"Average excluding first three values: {avg:.2f}")

    except ValueError as ve:
        print(f"ValueError: {ve}")
    except TypeError as te:
        print(f"TypeError: {te}")
    except Exception as e:
        print(f"Unexpected error: {e}")

main()
```

```python
class Greeting:
    def __init__(self, first, last):
        self.first = first
        self.last = last

    def display(self, formal):
        if formal:
            print(f"{self.last}, {self.first}")
        else:
            print(self.first)

    def __str__(self):
        return f"Greeting(first='{self.first}', last='{self.last}')"

g1 = Greeting("Mike", "Jones")

g1.display(True)
g1.display(False)
print(g1)
```

```python
class Student:
    def __init__(self, name: str, number: int, courses: list):
        self.name = name
        self.number = number
        self.courses = courses

    def get_name(self):
        return self.name

    def get_courses(self):
        return self.courses

    def set_name(self, new_name: str):
        self.name = new_name

    def __len__(self):
        return len(self.courses)

    def enroll(self, course):
        if course not in self.courses:
            self.courses.append(course)
```

```python
s1 = Student("Gabin", 101, ["Programming", "English"])

print(s1.get_name())
print(s1.get_courses())

s1.set_name("Gavin")
print(s1.get_name())

s1.enroll("Python")
print(s1.get_courses())

s1.enroll("Math")
print(s1.get_courses())

print(len(s1))
```

```python
class Point:
  def __init__(self, x, y):
    self.x = x
    self.y = y

  def distance(self, other):
    dx = (self.x - other.x) ** 2
    dy = (self.y - other.y) ** 2
    return (dx + dy) ** 0.5

class Rectangle:
  def __init__(self):
    self.p1 = Point(0,0)
    self.p2 = Point(5,0)
    self.p3 = Point(0,5)
    self.p4 = Point(5,5)
    self.width = self.p1.distance(self.p2)
    self.height = self.p1.distance(self.p3)

  def area(self):
    return self.width * self.height
```

```python
class Employee:
    def __init__(self, name, wage, hours):
        self.name = name
        self.wage = wage
        self.hours = hours

    def calculate_pay(self):
        return self.wage * self.hours

    def __str__(self):
        return f"{self.name}: wage={self.wage}, hours={self.hours}"

    def get_wage(self):
        return self.wage

    def set_wage(self, new_wage):
        self.wage = new_wage


class Manager(Employee):
    def __init__(self, name, wage, hours, bonus):
        super().__init__(name, wage, hours)
        self.bonus = bonus

    def calculate_pay(self):
        return (1 + self.bonus) * self.get_wage() * self.hours


employee = Employee("Bob", 50, 10)
manager = Manager("Alica", 100, 15, 0.2)

print(employee)
print(manager)
print()
print(f"Employee {employee.name} weekly payment is {employee.calculate_pay()}$")
print(f"Manager {manager.name} weekly payment is {manager.calculate_pay()}$ that includes a {manager.bonus * 100}% bonus.")
```
