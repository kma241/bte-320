```python
Sales = float(input("enter sales ($):"))
COGS = 0.4 * Sales
gross_profit = Sales - COGS
tax_rate = float(input("Enter tax rate ($): "))
taxes = gross_profit * tax_rate / 100
net = gross_profit - taxes

print(f"Sales: {Sales}$")
print(f"Cost of Goods Sold: {COGS}$")
print(f"Gross Prfoit: {gross_profit}$")
print(f"Taxes: {taxes}$")
print(f"Net Income: {net}$")
```

```python
num1 = float(input("Give me the first number: "))
num2 = float(input("Give me the second number: "))
operator = input("Enter an operation: ")

if operator == "+":
  print(num1 + num2)
elif operator == "-":
  print(num1 - num2)
elif operator == "*":
  print(num1 * num2)
elif operator == "/":
  print(num1/num2)
else:
  print("Invalid Operator")
```

```python
import random

play = input("Do you want to play a game(y/n)?")

while play == "y":
  player = input("Choose b/w rock, paper, scissors").lower()
  computer = random.choice(['rock', 'paper', 'scissors'])

  print(f"Player: {player} Computer: {computer}")

  if player == computer:
    print("Draw")
  elif player == 'rock' and computer == 'scissors':
    print("Player wins!")
  elif player == 'scissors' and computer == 'paper':
    print("Player wins!")
  elif player == 'paper' and computer == 'rock':
    print("Player wins!")
  else:
    print("Computer wins!")

  play = input("Do you want to play again(y/n)?")
```
